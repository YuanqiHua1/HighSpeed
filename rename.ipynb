```python
import  os

folder = r'U:\YuanqiHua\High speed\260127 dmrt3 MTZ\pkl_output'

for file in os.listdir(folder):
    name,ext = os.path.splitext(file)
    new_name = name[:-2]+ext
    older = os.path.join(folder,file)
    new_path = os.path.join(folder,new_name)
    os.rename(older,new_path)
    print(f"Renaming: {file} -> {new_name}")
```
